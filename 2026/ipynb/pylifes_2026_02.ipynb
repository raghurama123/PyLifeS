```python
import numpy as np
```

```python
A=[1,2,3]; A=np.array(A) # same as A=np.array([1,2,3]), A is a 1D array
```

```python
print(A.shape)
```

```python
element_first=A[0]; element_last=A[-1]; print(element_first, element_last)
```

```python
A=np.array([[1,2,3],[4,5,6]]); print(A); print(A.shape) # A is a 2D array; 2x3 matrix
```

```python
row_0=A[0,:]; print(row_0)
```

```python
col_1=A[:,1]; print(col_1)
```

```python
Nrow=A.shape[0]; Ncol=A.shape[1]; print(Nrow, Ncol)
```

```python
A=np.array([2,3,1,3,1]); print(A)
```

```python
print(A[:])      # all elements (full slice; start and stop are default)
```

```python
print(A[::+1])   # all elements with explicit step = +1 (same as A[:])
```

```python
print(A[::-1])   # all elements in reverse order (step = -1)
```

```python
A_sorted_asc=np.sort(A); print(A_sorted_asc)
```

```python
A_sorted_desc=A_sorted_asc[::-1]; print(A_sorted_desc) # [::-1] means all row, all col
```

```python
index_A_asc=np.argsort(A); print(index_A_asc)
```

```python
print('Minimum value of A is:',A_sorted_asc[0])

index_small=index_A_asc[0]
print('Minimum value of A is:',A[index_small],' with index ',index_small)
```

```python
print('Maximum value of A is:',A_sorted_asc[-1])
print('Maximum value of A is:',A_sorted_desc[0])

index_big=index_A_asc[-1] 
print('Maximum value of A is:',A[index_big],' with index ',index_big)
```

```python
index_A_desc=index_A_asc[::-1]; print(index_A_desc)
```

```python
index_big=index_A_desc[0] 
print('Maximum value of A is:',A[index_big],' with index ',index_big)
```
