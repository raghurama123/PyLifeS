## Assignment and Print statements

```python
x = 22/7                                  # This is an assignment statement

print(x)                                  # Print statement
```

```python
var = 2323
Var = 23233344
VAR = 232

print(var, Var, VAR)
```

```python
print("The value of x is:", x)            # Print statement with some text
print("The value of x is:", round(x, 7))  # Round-off x to 7 decimal places for printing
print(f"The value of x is: {x:15.7f}")    # Print x in 15 width and use 7 spaces for decimal part
```

```python
print(f"The value of x is: {1.1:15.7f}")  
print(f"The value of x is: {1000.3:15.7f}")  
```

```python
print(1.1,1000.3)
```

```python
x = 22/7
y = 7/22                        

print(x, y)
```

```python
x = 22/7; y = 7/22                        # Two statements separated by semi-colon

print(x, y)
```

```python
x, y, z = 22/7, 7/22, 22/7 + 7/22         # Multiple assignments in one line

print(f"x is: {x:.7f}, y is: {y:.7f}, z is: {z:.7e}") 
```

```python
x = 3E-2
y = 3*10**-2
print(x,y)
```

```python
x=1234.456789456789
print(f"{x:.6e}") # 1.23... * 10**3
```

## Math operations

```python
# Addition
stock_volume = 50 # volume in micro L
water_volume = 150 
total_volume = stock_volume + water_volume
print(total_volume) 
```

```python
StockVolume=20
print(StockVolume)
```

```python
# Subtraction
final_cells = 20000 
initial_cells = 40
net_increase = final_cells - initial_cells
print(net_increase) 
```

```python
# Multiplication
cells_per_ml = 2000
volume_ml = 100 
total_cells = cells_per_ml * volume_ml 
print(total_cells) 
```

```python
# Division
colonies = 420
dilution = 0.01
cells_per_ml = colonies / dilution
print(cells_per_ml)
```

```python
# Exponentiation
initial_cells = 2000
growth_factor = 10
generations = 3
final_cells = initial_cells * (growth_factor**generations)
print(final_cells) 
```

```python
# Floor division
time_hours = 10
generation_time = 3
generations = time_hours // generation_time
print(generations)
```

```python
# Modulus
samples = 23
tubes = 5
leftover = samples % tubes
print(leftover) 
```

## Datatypes

```python
a = 1; b = 1.0; c = complex(1,0); d = 'One'

type1=type(a); type2 = type(b); type3 = type(c); type4 = type(d)

print(type1, type2, type3, type4)
```

```python
params={'mu':1.0, 'sigma':0.5}
print(params, type(params))
```

```python
print(len(params))
print(params.keys())
print(params.values())
```

```python
params.update({'scal':1.0}) 
print(params)
```

```python
params.update({'offset':0.0,'center':[1,2,3]}) 
print(params)
```

```python
a=1
type1=type(a)

a=1.0
type2=type(a)

a=complex(1,0) 
# Note: For the complex number, you have to specify real and imaginary parts inside a parenthesis.
type3=type(a)

a='ONE is 1'  
#Note: a string is specified inside single quotes.
type4=type(a) 

print(type1,type2, type3, type4)
```
