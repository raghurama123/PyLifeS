# Multiple columns

```python
data={
    'Species':['E.coli', 'Yeast', 'Zebrafish'],
    'Doubling time (hrs)': [0.5, 1.5, 48],
    'Temperature (C)': [37, 30, 28]
}

df=pd.DataFrame(data); print(df)
```

```python
# header of each column (or column name)
columns=['Species', 'Doubling time (hrs)', 'Temperature (C)']
# column-1
species=['E.coli', 'Yeast', 'Zebrafish']
# column-2
doubling_time=[0.5, 1.5, 48]
# column-3
temperature=[37, 30, 28]

data=dict(zip(columns, [species, doubling_time, temperature]))

df=pd.DataFrame(data); print(df)
```

```python
df.iloc[0] # first row in the DataFrame
```

```python
df.iloc[0,1] # first row and second column
```
